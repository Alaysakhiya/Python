```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

data = pd.read_csv(r"D:\Python\Project\Final_Exam\2019.csv")


print("\n\t======Global Happiness Report Analysis=====")

while True:

        

    print("\n1. Histogram")
    print("2. Bar Chart Top 10")
    print("3. Bar Chart Bottom 10")
    print("4. Scatter Chart")
    print("5. Exit")
    choice = int(input("\nEnter your Choice :> "))

    if choice == 1:

        print("\n-----Hist Ploat-----")
        sb.histplot(data=data ,x="Healthy life expectancy")
        plt.title("Happiness Score Count") 
        plt.xlabel("Healthy life expectancy")
        plt.ylabel("Country Count")
        plt.tight_layout()
        plt.show()
        print("\nDisplay Successfully !")

    elif choice == 2:

        print("\n-----Bar Ploat-----")
        plt.figure(figsize=(9,5))
        sb.barplot(data=data.head(10),x="Country", y="Perceptions of corruption",hue="Country")

        plt.title("Corruption Protection on Top 10 Country")
        plt.xticks(fontsize=7.5)
        plt.xlabel("Country")
        plt.ylabel("Perceptions of Corruption")
        plt.show()
        print("\nDisplay Successfully !")

    elif choice == 3:

        print("\n-----Bar Ploat-----")
        plt.figure(figsize=(10,5))
        sb.barplot(data=data.tail(10),x="Country", y="Perceptions of corruption",hue="Country")
        plt.xticks(fontsize=6)
        plt.title("Corruption Protection on Bottom 10 Country")
        plt.xlabel("Country")
        plt.ylabel("Perceptions of Corruption")
        plt.show()
        print("\nDisplay Successfully !")


    elif choice == 4:

        print("\n-----Scatter Ploat-----")
        plt.figure(figsize=(8,5))
        sb.scatterplot(data=data,x="GDP per capita" ,y="Healthy life expectancy")
        plt.title("GDP vs Hapiness Score")
        plt.xlabel("GDP per capita")
        plt.ylabel("Happiness Score")
        plt.grid(True)
        plt.show()
        print("\nDisplay Successfully !")

    elif choice == 5:
        print("\nThank you for Visiting")
        break
    else:
        print("Invalid Choice !")
```
